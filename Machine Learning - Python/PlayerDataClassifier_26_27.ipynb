**Parameter Tuning**

The accuracy should be good, but can it be better?

Most machine learning algorithms have parameters that can be tuned to improve performance.

A common approach for testing is 'GridSearch' where various combinations of parameters are trialled to see which is best.

The cell below sets the parameters that will be varied, and the values for each. Each configuration will then be tested in turn.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': range(1, 10, 1), # start, stop and then the step
    'min_samples_leaf': range(1, 20, 2),
    'min_samples_split': range(2, 20, 2)
}

tree = DecisionTreeClassifier(random_state=1)

grid_search = GridSearchCV(estimator=tree, param_grid=param_grid,
                           cv=5, verbose=True)
grid_search.fit(X_train, y_train)

```

Run the below cell to print the best configuration, and its accuracy. Was performance improved?



```python
print("best accuracy", grid_search.best_score_)
print(grid_search.best_estimator_)
```

**Further Performance Improvement**

There are several more advanced machine learning algorithms. One is [LightGBM](https://github.com/lightgbm-org/LightGBM), a tree based learning algorithms that uses the concept of is a gradient boosting to improve performance.

Full details are given at the link for your further research.

For now, we can use this library to train another model, ready for importing into the C++ game.

Run the below cells to define and train the new model.

```python
import lightgbm as lgb
```

```python
model = lgb.LGBMClassifier(
    objective="multiclass",
    n_estimators=100,
    learning_rate=0.1,
    num_leaves=8,
    min_child_samples=5,
    random_state=42,
    verbose=-1,
)
```

```python
model.fit(
    X_train, y_train,
    eval_set=[(X_test, y_test)],
    callbacks=[lgb.early_stopping(10, verbose=False)],
)
```

**Feature Importance**

Another concept we explored in the lecture is the challenge of explaining the machine learning model, or understanding its decision making.

In this simple example, we trained on three features. Is one feature more important to the classification than others? One way to epxlore is by looking at Feature Importance.

Run the below cells to plot charts showing the overall importance of the Features to the model.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
lgb.plot_importance(model, importance_type="split", ax=axes[0],
                    title="Feature importance (split count)")
lgb.plot_importance(model, importance_type="gain", ax=axes[1],
                    title="Feature importance (gain)", precision=1)
```
